```python
import seaborn as sns
import pandas as pd
df = sns.load_dataset('iris')

print("The first 10 rows are : ")
print(df.head(10))

print("Missing values per column are : ")
print(df.isnull().sum())

df["petal_ratio"] = df["petal_length"] / df["petal_width"]

print("Dataframe with the new petal_ratio column : ")
print(df[["petal_length", "petal_width", "petal_ratio"]].head())
```
